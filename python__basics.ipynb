```python
##Calculator
num1 = float(input("Num1 :"))
num2 = float(input("Num2 :"))
operation = input("Enter the operation needs to be performed: (+,-,*,/): ")
if (operation == "+"):
    print("sum =", num1+num2)
elif(operation == "-"):
    print(num1-num2)
elif(operation == "*"):
    print(num1*num2)
elif(operation =="/"):
    if(num2==0):
        print("Zero division Error")
    else:
        print(num1/num2)
else:
    print("Uknown operation", operation)

    
```

```python
## grading
grade=float(input("Enter your grade /20: "))
if(grade > 20 and grade < 0):
    print("invalid entry of grade")
elif(grade >=16):
    print("very good")
elif(grade< 16 and grade >= 14):
    print("good")
elif(grade < 14 and grade>=12):
    print("fairly good")
elif(grade <12 and grade >=10):
    print("Pass")
else:
    print("fail")
```

```python
## Guessing game
import random
secret = random.randint(1,100)
guess = int(input("Guess a number:"))
while guess!= secret:
    if(guess < secret):
        print("Too Low")
    elif(guess> secret):
        print("Too  High")
    guess= int(input("Guess another time:"))
print("bravo you found the secret:", secret)
```

```python
## Guessing game
import random
secret = random.randint(1,100)
attempts=0
while attempts < 7:
    guess = int(input("Guess a number:"))
    attempts=attempts+1
    if(guess == secret):
        print("bravo you found the secret:", secret, "in attempt", attempts)
        break
    elif(guess < secret):
        print("Too Low")
    elif(guess> secret):
        print("Too  High")
if(guess != secret):
    print("you Lost, the secret was", secret)
```

```python
##sum of 1 to 5(while)
total = 0
i=1
while i<6:
    total= total+i
    i=i+1
print(total)
```

```python
#ENUMURATE
scores =[22,34,45,32,67]
best=-1
posi=0
for pos,s in enumerate(scores, start=1):
    if s>best:
        best=s
        posi=pos
print("Best score:",best,"student:",posi)
```

```python
answers =["b","a","c","d","e"]
correct= ["b","c","a","d","e"]
scores=0
for given,right in zip(answers,correct):
    if given == right:
        scores=scores+1
print(scores,"/5")
```
