```python
##Calculator
num1 = float(input("Num1 :"))
num2 = float(input("Num2 :"))
operation = input("Enter the operation needs to be performed: (+,-,*,/): ")
if (operation == "+"):
    print("sum =", num1+num2)
elif(operation == "-"):
    print(num1-num2)
elif(operation == "*"):
    print(num1*num2)
elif(operation =="/"):
    if(num2==0):
        print("Zero division Error")
    else:
        print(num1/num2)
else:
    print("Uknown operation", operation)

    
```

```python
## Guessing game
import random
secret = random.randint(1,100)
guess = int(input("Guess a number:"))
while guess!= secret:
    if(guess < secret):
        print("Too Low")
    elif(guess> secret):
        print("Too  High")
    guess= int(input("Guess another time:"))
print("bravo you found the secret:", secret)
```

```python
## Guessing game
import random
secret = random.randint(1,100)
attempts=0
while attempts < 7:
    guess = int(input("Guess a number:"))
    attempts=attempts+1
    if(guess == secret):
        print("bravo you found the secret:", secret, "in attempt", attempts)
        break
    elif(guess < secret):
        print("Too Low")
    elif(guess> secret):
        print("Too  High")
if(guess != secret):
    print("you Lost, the secret was", secret)
```

```python
contacts={}
while True:
    choice = input("1 add, 2 search, 3 list, 0 quit: ")
    if(choice == "1"):
        name=input("Name: ")
        phone=input("Phone: ")
        if name not in contacts:
            contacts[name]=[]
        contacts[name].append(phone)
    elif choice =="2":
        name=input("Name: ")
        print(contacts.get(name, "Not Found"))
    elif choice =="3":
        for name in sorted(contacts):
            print(f"{name:<5}",",".join(contacts[name]))
    elif choice =="0":
        break

        
```

```python
counts = {}
with open("experiment.txt", encoding="utf-8") as f:
    for line in f:
        line = line.lower()
        for p in ",.!?":
            line = line.replace(p, "")
        for word in line.split():
            counts[word] = counts.get(word, 0) + 1

for word, n in counts.items():
    print(f"{word:<10}{n}")
```

```python
with open("experiment.txt", encoding="utf-8") as f:
    text = f.read().lower()

for p in ",.!?":
    text = text.replace(p, "")

words = text.split()

counts = {}
for w in words:
    counts[w] = counts.get(w, 0) + 1

with open("report.txt", "w", encoding="utf-8") as out:
    out.write(f"Words: {len(words)}\n")
    out.write(f"Unique: {len(counts)}\n")
    for w in sorted(counts):
        out.write(f"{w:<10}{counts[w]}\n")

print("Report saved")
```
